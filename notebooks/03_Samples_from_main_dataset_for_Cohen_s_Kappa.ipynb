```python
INPUT_PATH = "/content/drive/MyDrive/ShopHallu/Annotator 1 (Main Dataset).csv"
OUTPUT_PATH = "/content/drive/MyDrive/ShopHallu/Samples_from_main_dataset_for_Cohen_s_Kappa.csv"
SAMPLE_SIZE = 500
RANDOM_SEED = 42
```

```python
df = pd.read_csv(INPUT_PATH)
print(f"Main dataset loaded: {len(df)} rows")
```

```python
sample_df = df.sample(n=SAMPLE_SIZE, random_state=RANDOM_SEED).reset_index(drop=True)
```

```python
columns_to_keep = [
    "asin",
    "category",
    "generation_reviews",
    "generation_text",
    "influencer_review",
    "evidence_review",
    "supported",
    "fabricated_feature",
    "exaggerated_performance",
    "contradiction",
    "Unsupported Promotional Implication",
    "annotator_comment"
]

sample_df = sample_df[columns_to_keep]

annotation_cols = [
    "supported",
    "fabricated_feature",
    "exaggerated_performance",
    "contradiction",
    "Unsupported Promotional Implication",
    "annotator_comment"
]

sample_df[annotation_cols] = ""

sample_df.to_csv(OUTPUT_PATH, index=False)
print(f"Sampled {len(sample_df)} rows -> saved to {OUTPUT_PATH}")
```
